```python
import numpy as np
import cv2 as cv
from matplotlib import pyplot as plt

img = cv.imread('/home/martin/turtlebot3_ws/src/lab1/ben.jpg')

assert img is not None, "file could not be read, check with os.path.exists()"

gray_img = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
plt.subplot(121),plt.imshow(img_rgb)
plt.title('Original Image'), plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(gray_img, cmap='grey')
plt.title('Greyscale Image'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
img = cv.imread('ben.jpg', cv.IMREAD_GRAYSCALE) 
assert img is not None, "file could not be read call for help"
edges = cv.Canny(img, 100, 200)

plt.subplot(121),plt.imshow(img,cmap = 'gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(edges,cmap = 'gray')
plt.title('Edge Image'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
import cv2 as cv
face_cascade = cv.CascadeClassifier(cv.data.haarcascades + 'haarcascade_frontalface_default.xml')

faces = face_cascade.detectMultiScale(gray_img, scaleFactor = 1.1, minNeighbors = 4)
for (x,y,w,h) in faces:
	cv.rectangle(img, (x,y), (x+w, y+h), (255,0,0), 2)
cv.imshow('face detect', img)
cv.waitKey(0)
cv.destroyAllWindows()
plt.subplot()
```
