```python
import numpy as np
from scipy.linalg import solve
```

```python

A = np.array([[3, 1], [1, 2]])
b = np.array([9, 8])
x = solve(A, b)
print(x)
```

```python
from scipy.optimize import minimize

#coeffs
x0 = np.array([0])
def y(x):
    return x**2 + 2*x
min_y = minimize(y, x0, method='nelder-mead')

print(min_y.x)
```

```python
from scipy.fft import fft, fftfreq
import matplotlib.pyplot as plt
def f(x):
    return np.sin(100 * np.pi * x) + 0.5 * np.sin(160 * np.pi * x)

# duration: 1 sec, N = 1000
t = np.linspace(0, 1, 1000, endpoint="false")

samples = f(t)

vals = fft(samples)
amplitude = np.abs(vals) * 1/500

freq = fftfreq(1000, 1/1000)
pos_freqs = freq > 0

freq = freq[pos_freqs]
amplitude = amplitude[pos_freqs]

#do we want the negative frequencies
fig, ax = plt.subplots(figsize=(5, 3), layout='constrained')
ax.plot(freq, amplitude)
ax.set_ylabel('amplitudes')
ax.set_xlabel('frequencies')
```
